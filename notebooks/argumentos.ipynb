## Cap 6 mutabilidad
====================

### Objetos inmutables

```python
a = 3
b = a
print("b es", b)
a = 'spam'
print("b es", b)
```

Los objetos integer son inmutables

```python
a = 3
b = a
a = a + 2
print("b es", b)
```

### Objetos mutables

#### Listas

```python
L = [2, 3, 4]
M = L
L[0] = 24
print("L es", L)
print("M es", M)
```

Copia de objetos

```python
L = [2,3,4]
M = L[:]
L[0] = 24
print("L es", L)
print("M es", M)
```

### Referencias e igualdades

- Operador _==_ igualdad de valores
- Operador _is_ identidad de objeto

```python
L = [1, 2, 3]
M = [1, 2, 3]
print("L == M", L == M)
print("L is M", L is M)
```

```python
L = [1, 2, 3]
L = M
L == M
L is M
```
