### Global Scope

- Global names are variables assigned at the top level of the enclosing module file.
- Global names must be declared only if they are assigned within a function.
- Global names may be referenced within a function without being declared.

```python
x = 88

def func():
    x = 99

func()
print(x)
```

```python
x = 88

def func():
    global x
    x = 99

func()
print(x)
```

### LEGB rule

```python
y, z = 1, 2

def all_global():
    global x
    x= y + z
all_global()
print('x = %d, y = %d, z = %d' % (x, y, z))
```

### Nested Scopes

```python
x = 99
def f1():
    x = 88
    def f2():
        print ("x local es =", x) #Sin coma no funciona (syntax error "perhaps you forgot a comma")
    f2()
f1()
print("x global es =", x)
```

### Factory Functions: Closures

A __closure__ or a __factory function__, the former describing a __functional programming technique__, and the latter denoting a __design pattern__. The function object in question remembers values in enclosing scopes regardless of whether those scopes are still present in memory.

```python
def f1():
    x=88
    def f2():
        print("x =", x)
    return f2
action = f1()
action()
```

```python
def maker(n):
    def action(x):
        return x ** n
    return action
f = maker(2)
f(3)
f(4)
    
```
